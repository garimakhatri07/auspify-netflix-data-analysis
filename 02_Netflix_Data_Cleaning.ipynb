Identify and handle missing values.

```python
# Check missing values in each column

missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values)

print("\nTotal Missing Values:", missing_values.sum())
```

```python
# Identify 'Not Given' values

for column in df.select_dtypes(include='object').columns:

    count = df[column].str.strip().str.lower().eq('not given').sum()

    if count > 0:
        print(f"{column}: {count} unavailable values")
```

```python
# Replace 'Not Given' with NaN

df.replace(
    to_replace=r'(?i)^\s*Not Given\s*$',
    value=np.nan,
    regex=True,
    inplace=True
)

print("Placeholder values handled successfully!")
```

```python
# Check missing values after replacement

missing_after = df.isnull().sum()

print("Missing Values After Cleaning:")
print(missing_after)

print("\nTotal Missing Values:", missing_after.sum())
```

```python
# Verify that no records were deleted

print("Dataset Shape After Missing Value Handling:", df.shape)
```

### Missing Value Analysis and Handling

The dataset was examined for missing values using Pandas. Initially, no actual null values or empty records were detected.

However, the Director and Country columns contained placeholder entries represented as "Not Given". These entries indicated unavailable information.

To handle this issue, the placeholder values were converted into standard missing values (NaN). The affected records were retained instead of being deleted because other information in those records remained useful for further analysis.

**Key Findings:**
- Director: 2,588 unavailable values.
- Country: 287 unavailable values.
- Total unavailable values identified: 2,875.
- No records were deleted during this process.
- Final dataset dimensions remained unchanged at 8,790 rows and 10 columns.

This approach preserves the dataset while making missing information easier to identify and handle during subsequent analysis.

Remove duplicate records and formatting inconsistencies.

```python
# Check duplicate records

duplicate_count = df.duplicated().sum()

print("Total Duplicate Records:", duplicate_count)
```

Check formatting inconsistencies

```python
# Check leading and trailing spaces in text columns

print("Formatting Inconsistency Check:\n")

for column in df.select_dtypes(include='object').columns:

    values = df[column].dropna().astype(str)

    inconsistent = (
        values != values.str.strip()
    ).sum()

    print(f"{column}: {inconsistent} inconsistent entries")
```

```python
# Create a copy before formatting corrections

df_before_format = df.copy()

# Clean whitespace in all text columns

for column in df.select_dtypes(include='object').columns:

    df[column] = (
        df[column]
        .str.replace('\u00A0', ' ', regex=False)
        .str.replace(r'\s+', ' ', regex=True)
        .str.strip()
    )

print("Text formatting cleaned successfully!")
```

```python
# Verify formatting corrections

print("Formatting Issues After Cleaning:\n")

for column in df.select_dtypes(include='object').columns:

    values = df[column].dropna().astype(str)

    inconsistent = (
        values != values.str.strip()
    ).sum()

    print(f"{column}: {inconsistent} inconsistent entries")
```

### Duplicate Records and Formatting Consistency

The dataset was examined for duplicate records using Pandas. The initial duplicate check identified zero exact duplicate rows, so no records required deletion.

Text columns were then inspected for formatting inconsistencies, including leading spaces, trailing spaces, repeated whitespace and non-breaking space characters.

A formatting issue was identified in the title column and corrected using string-cleaning operations. The cleaning process preserved the original information while improving text consistency.

**Key Findings:**
- Total duplicate records identified: 0.
- Duplicate records removed: 0.
- Unnecessary whitespace corrected in the title column.
- Text formatting was standardized across object columns.
- All 8,790 records were retained.

These checks help maintain data consistency and ensure the dataset is suitable for further analysis.
